```python
from datasets import load_dataset

ds = load_dataset("stanfordnlp/imdb")
ds
```

# converting to pandas

```python
import pandas as pd

df = pd.DataFrame(ds["train"])
df
```

```python
print(df.shape)
print(df.info())
print(df["label"].value_counts())
```

```python
df.head()
```

```python
print(df["label"].value_counts())
import matplotlib.pyplot as plt

df["label"].value_counts().sort_index().plot(
    kind="bar",
    xlabel="label",
    ylabel="Number of reviews",
    title="IMDB Dataset Class Distribution"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
# Length in words
df["word_length"] = df["text"].str.split().str.len()

print(df[["text","word_length"]].head())
print(df["word_length"].describe())

shortest = df.loc[df["word_length"].idxmin()]
longest = df.loc[df["word_length"].idxmax()]

print("Shortest review:", shortest["word_length"], "words")
print("Longest review:", longest["word_length"], "words")


df["word_length"].hist(bins=50)

plt.xlabel("Number of words")
plt.ylabel("Number of reviews")
plt.title("Distribution of Review Lengths")
plt.show()
```

```python
# Display 5 sample reviews from the training dataset
print(ds["train"].to_pandas().head())
```

```python
#display of few sample reviews from the training datasets
# Display 5 sample reviews
for i in range(5):
    review = ds["train"][i]["text"]
    label = ds["train"][i]["label"]

    sentiment = "Positive" if label == 1 else "Negative"

    print(f"Review {i + 1}: {review}")
    print(f"Sentiment: {sentiment}")
    print("-" * 80)
```

```python
#cleaning the dataset

from sklearn.model_selection import train_test_split

# Split the df data into 80% training and 20% validation sets
# We set a random_state to make sure the split is reproducible
train_split, val_split = train_test_split(df, test_size=0.2, random_state=42)

# Display the sizes of the splits to confirm
print(f"Original training size: {len(df)}")
print(f"New training split size: {len(train_split)}")
print(f"Validation split size: {len(val_split)}")
```
